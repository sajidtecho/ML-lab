# End-to-End Exploratory Data Analysis (EDA) on Iris Dataset (Kaggle Dataset)

In this notebook, we perform a comprehensive end-to-end Exploratory Data Analysis (EDA) on the classic **Iris Dataset** downloaded dynamically using the `kagglehub` library.

```python
# 1. Import necessary libraries for data analysis and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import kagglehub
import os
```

```python
# 2. Download latest version of the Iris dataset from Kaggle using kagglehub
path = kagglehub.dataset_download("uciml/iris")
print("Path to dataset files:", path)
```

```python
# 3. Load the Iris.csv file into a Pandas DataFrame
csv_path = os.path.join(path, "Iris.csv")
df = pd.read_csv(csv_path)
```

```python
# 6. Display the first 5 rows of the loaded dataset
df.head()
```

```python
# 7. Check the dimensions (shape) of the dataset
df.shape
```

```python
# 8. Check columns, data types, and non-null values
df.info()
```

```python
# 9. Verify if there are any missing/null values in any of the columns
df.isnull().sum()
```

```python
# 10. Generate descriptive statistical summary for the numerical features
df.describe()
```

```python
# 11. Check the count of each species class to ensure the dataset is balanced
# Fix the column name mismatch: dataset has "Species", notebook uses "species"
if 'Species' in df.columns and 'species' not in df.columns:
    df.rename(columns={'Species': 'species'}, inplace=True)

df['species'].value_counts()
```

```python
# 12. Plot histograms of all numerical features to examine their distribution characteristics
df.drop(columns=['species']).hist(figsize=(10, 8), bins=15, edgecolor='black')
plt.suptitle('Histograms of Iris Numerical Features (Kaggle Dataset)', fontsize=14)
plt.show()
```

```python
# 13. Plot boxplots to see the range, quartiles, and outliers for each feature grouped by species
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
feature_cols = ['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm']
for i, feature in enumerate(feature_cols):
    row, col = i // 2, i % 2
    sns.boxplot(ax=axes[row, col], x='species', y=feature, data=df, palette='Set2')
    axes[row, col].set_title(f'Distribution of {feature} by Species')
plt.tight_layout()
plt.show()
```

```python
# 14. Display a pairplot to inspect pairwise relationships and class separability
sns.pairplot(df, hue='species', palette='Set1', height=2.5)
plt.show()
```

```python
# 15. Plot violin plots to display the full distribution density shape for each feature across species
fig, axes = plt.subplots(2, 2, figsize=(12, 10))
for i, feature in enumerate(feature_cols):
    row, col = i // 2, i % 2
    sns.violinplot(ax=axes[row, col], x='species', y=feature, data=df, palette='muted')
    axes[row, col].set_title(f'Violin Plot of {feature} by Species')
plt.tight_layout()
plt.show()
```

```python
# 16. Compute the correlation matrix for the numerical features
correlation_matrix = df.drop(columns=['species']).corr()
```

```python
# 17. Plot the correlation matrix heatmap to check for multi-collinearity
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix Heatmap', fontsize=14)
plt.show()
```
